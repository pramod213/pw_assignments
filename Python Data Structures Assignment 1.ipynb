# Q2. Given are some following variables containing data: What will be the data type of the above given variable.

```python
# Given variables
var1 = ' '                      # (i)
var2 = '[ DS , ML , Python]'   # (ii)
var3 = ['DS', 'ML', 'Python']  # (iii)
var4 = 1                       # (iv)
```

## Solution:

```python
print(type(var1))  # <class 'str'>
print(type(var2))  # <class 'str'>
print(type(var3))  # <class 'list'>
print(type(var4))  # <class 'int'>
```

### Solution:
(i) var1 = ‘ ‘  ---> Type: str (string)    
                      Reason: It's a string containing a single space character.

(ii) var2 = ‘[ DS , ML , Python]’ ---->  Type: str (string) 

Reason: It looks like a list, but it's actually a string representation of a list. It's enclosed in quotes.

(iii) var3 = [ ‘DS’ , ’ML’ , ‘Python’ ]  ---> Type: list

Reason: This is an actual Python list containing three strings.



(iv) var4 = 1.  ---->  Type: int (integer)

Reason: A whole number with no decimal places.


# Q4. Create a list of length 10 of your choice containing multiple types of data. Using for loop print the element and its data type.

```python
#Solution:
# List with multiple data types
my_list = [
    "Hello",          # string
    42,               # integer
    3.14,             # float
    True,             # boolean
    None,             # NoneType
    [1, 2, 3],        # list
    (4, 5),           # tuple
    {'a': 1},         # dictionary
    {1, 2, 3},        # set
    b"bytes"          # bytes
]

# Loop through the list 
for item in my_list:
    print(f"Value: {item}, Type: {type(item)}")
```

# Q5. Using a while loop, verify if the number A is purely divisible by number B and if so then how many times it can be divisible.

```python
# Solution: 
# Example values for A and B
A = 64
B = 2

count = 0

while A % B == 0:
    A = A // B  
    count += 1

print(f"The number is divisible {count} times by {B}")
```

# Q6. Create a list containing 25 int type data. Using for loop and if-else condition print if the element is divisible by 3 or not.

```python
# Solution:
# List of 25 integers
numbers = list(range(1, 26))  # This creates a list from 1 to 25

# Loop through each number and check if it's divisible by 3
for num in numbers:
    if num % 3 == 0:
        print(f"{num} is divisible by 3")
    else:
        print(f"{num} is NOT divisible by 3")
```

# Q7. What do you understand about mutable and immutable data types? Give examples for both showing this property.

## Solution: 
### Mutable Data Types
Definition: Mutable objects can be changed after creation.

You can modify their content (like add, update, or delete elements) without changing their identity (memory address).

Examples of Mutable Data Types in Python:

list

dict

set

```python
# Example of Mutable Data Types
my_list = [1, 2, 3]
print("Before:", my_list)

my_list.append(4)  # After Modifying the list
print("After:", my_list)
```

## Immutable Data Types
Definition:   Immutable objects cannot be changed after creation.
Any change creates a new object with a new memory address.

Examples of Immutable Data Types in Python:   int , float , str and tuple .

```python
# Example of Immutable Data Types
my_str = "hello"
print("Before:", my_str)

my_str = my_str + " world"  # This creates a new string object
print("After:", my_str)
```
